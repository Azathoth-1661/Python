串列 List — 認識與基本操作

串列是一種可以儲存多個元素的容器，使用 [ ] 括號。

fruits = ["apple", "banana", "cherry"]
numbers = [10, 20, 30, 40]
串列的特性：

有順序（可透過索引存取）/ 可修改（mutable）/ 可包含不同型別的資料

```python
fruits = ["apple", "banana", "cherry"]
numbers = [10, 20, 30, 40]
```

存取元素	fruits[0] → "apple"
修改元素	fruits[1] = "orange"
串列長度	len(fruits)
加入元素	fruits.append("kiwi")
插入元素	fruits.insert(1, "mango")
刪除元素	fruits.remove("banana") 或 del fruits[0]
清空串列	fruits.clear()

```python
fruits = ["apple", "banana", "cherry"]
numbers = [10, 20, 30, 40]

for fruit in fruits:
    print(fruit)
```

```python
for i in range(len(fruits)):
    print(i, fruits[i])
```

```python
for i, fruit in enumerate(fruits):
    print(i, fruit)
```

# 常用方法
append(x)	將 x 加入到最後
insert(i, x)	在位置 i 插入 x
remove(x)	刪除第一個出現的 x
pop()	移除並回傳最後一個元素
sort()	就地排序
reverse()	反轉串列順序
index(x)	回傳 x 的索引位置
count(x)	回傳 x 出現次數

len() 函數簡介
回傳「容器型別」資料（例如：字串、串列、集合、字典）的元素數量。
len() 是「函數」而不是「方法」。
不適用於數字（int / float），例如 len(123) 會報錯。

```python
fruits = ["apple", "banana", "cherry"]
print(len(fruits))  # 輸出：3

text = "hello world"
print(len(text))  # 輸出：11（包含空格）

student = {"name": "John", "age": 18}
print(len(student))  # 輸出：2（有兩個鍵）

empty_list = []
print(len(empty_list))  # 輸出：0
```

```python
names = ["Amy", "Bob", "Charlie"]
print(len(names))
```

1. 串列的新增（append, insert）

```python
fruits = ["apple", "banana"]

# 使用 append() 將元素加在尾端
fruits.append("cherry")  # ["apple", "banana", "cherry"]
print(fruits)
# 使用 insert() 在指定位置插入元素
fruits.insert(1, "grape")  # ["apple", "grape", "banana", "cherry"]
print(fruits)
```

✅ 小挑戰：建立一個空串列，請使用者輸入三個水果名稱並加進串列中。

```python
fruits = []
count = 0
while count < 3:
    new_fruit = input("請輸入水果名稱：")
    if new_fruit in fruits:
        print("你已經輸入過這個水果了！請重新輸入。")
        continue  # 跳過這次迴圈，重新輸入
    fruits.append(new_fruit)
    count += 1
print("你輸入的水果有：", fruits)
```

in 的基本用途：判斷某個值是否存在於某個容器（像是字串、串列、集合等)

```python
fruits = ["apple", "banana", "cherry"]
# 判斷元素是否存在於串列中
print("apple" in fruits)  # True
print("grape" in fruits)  # False
# 判斷字元是否在字串中
word = "hello"
print("e" in word)     # True
print("z" in word)     # False
# 判斷元素是否在集合（set）中
colors = {"red", "green", "blue"}
print("red" in colors)    # True
print("yellow" in colors) # False
#配合 not in 判斷「不在」某集合中
if "apple" not in fruits:
    print("你還沒輸入 apple！")
#配合 for 迴圈使用 —「逐一取出元素」
for fruit in fruits:
    print(fruit)

```

說明：
for X in Y 是 Python 最常見的迴圈寫法
每次迴圈會把 Y（容器）裡的每個元素，依序指定給變數 X。
