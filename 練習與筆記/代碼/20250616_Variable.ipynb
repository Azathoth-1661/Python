其他種類的轉換

```python
a = 100
b = 123.45
c = '123'
d = '100'
e = '123.45'
f = 'hello, world'
g = True
print(int(c, base=16))  # str类型的'123'按十六进制转成int，输出291
print(int(d, base=2))   # str类型的'100'按二进制转成int，输出4
print(bool(f))       # str类型的'hello, world'转成bool，输出True
print(int(g))       # bool类型的True转成int，输出1
print(chr(a))       # int类型的100转成str，输出'd'
print(ord('d'))      # str类型的'd'转成int，输出100
```

使用input()鍵盤輸入函數完成基礎運算

「%」的字串格式化是「舊式」的格式化方法，適用於 Python 2 和 3 的版本，操作方式為「格式化字串 % 資料」，輸出結果會將資料插入格式化字串的位置。

%s	字串

%d	十進制整數

%x	十六進制整數

%o	八進制整數

%b	二進制整數

%f	十進制浮點數

%e	指數浮點數

%g	十進制或指數浮點數

%%	常值 %


```python
a = int(input("請輸入a的值："))
b = int(input("請輸入b的值："))
print("%d + %d = %d" %(a, b, a + b))
print("%d - %d = %d" %(a, b, a - b))
print("%d * %d = %d" %(a, b, a * b))
print("%d * %d = %d" %(a, b, a ** b)) #10*10的五次方
print("%d * %d = %d" %(a, b, a / b))
print("%d * %d = %d" %(a, b, a // b)) #取整除法 返回兩個數相除的商，且去除小數部分，只保留整數部分。
print("%d * %d = %d" %(a, b, a % b))  #a/b 的餘數
```

華氏溫度(F)轉攝氏溫度(C)

```python
f = input("請輸入華氏溫度：")
c = (float(f) - 32) * 5 / 9
print("攝氏溫度為：%.1f" %(c)) #.1f = 取小數點後一位 .2f取兩位
```

攝氏溫度轉華氏溫度

```python
c = input("請輸入攝氏溫度：")
f = float(c) * 9 / 5 + 32
print("華氏溫度為：%.1f" %(f))
```

使用圓的半徑計算周長與面積

使用 math 模組來取得更精確的 π（Pi）值，這樣更符合工程與科學的計算標準

```python
import math
radius = float(input("請輸入圓的半徑："))
perimeter = 2 * math.pi * radius
area = math.pi * radius * radius
print("圓周長為：%.2f" %(perimeter))
print("圓面積為：%.2f" %(area))
```

練習題2. 資料型態轉換與加總 並且利用 try/except來進行防呆

```python
try:
    a = int(input("請輸入第一個數字："))
    b = int(input("請輸入第二個數字："))
    print("總和是：", a + b)
except ValueError:
    print("請輸入有效的整數！")
```

輸入一個數字 判斷為基數或偶數

```python
try:
    int_1 = int(input("請輸入整數："))
    if int_1 % 2 == 0:
        print("是偶數")
    else:
        print("是奇數")
except ValueError:
    print("請輸入有效的整數！")
```
