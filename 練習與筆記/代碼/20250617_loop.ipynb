range參數 是一個「用來產生一串連續數字」的內建函式

它會「產生一組數列」，但有幾個重點:
起始值預設是 0
結束值「不包含」自己（不包含尾端）
間隔（step）預設是 1，可為負數

range(n)	產生 0 ~ n-1
range(a, b)	產生 a ~ b-1
range(a, b, s)	a ~ b-1，每次間隔 s

練習 2：輸入密碼直到正確
請使用者反覆輸入密碼，直到輸入正確（例如 "pass123"）。

🎯 目標：while + 字串比對 + 無限迴圈 + break

```python
password = "pass123"

while True:
    password_1 = input("請輸入密碼：")
    if password_1 == password:
        print("密碼正確，登入成功！")
        break
    else:
        print("密碼錯誤，請再試一次。")
```

練習 2進階版 錯誤三次鎖定：輸入密碼直到正確

```python
password = "pass123"
attempts = 3
while True:
    password_1 = input("請輸入密碼：")
    if password_1 == password:
       print("密碼正確，登入成功！")
       break
    else:
        attempts -= 1
        if attempts >0:
           print(f"密碼錯誤，你還剩下{attempts}次機會")
        else:
           print("帳號鎖定，請聯絡管理員。")
           break #如果不加入 會導致計數為負 跳不出迴圈
```

從 1 開始加總，直到總和超過 100 為止
請使用 while 迴圈，從 1 開始依序加總，直到總和超過 100 為止。
你需要最後印出：
1.總和是多少？
2.最後加到第幾個數字？

```python
i_number = 1
i_total = 0
while i_number < 101:
    i_total += i_number
    i_number +=1
print(f"總和為{i_total},加總倒第{i_number-1}個數字")    
```

 練習題 1：只加偶數的總和
請使用 while 迴圈計算 1 到 100 之間的所有偶數的總和，最後輸出總和與總共加了幾個偶數數字。

提示：

偶數的條件是 i % 2 == 0

你需要追蹤總和與加了幾次（像之前那樣）

```python
i_number = 1 #第一個數字 為1 
i_total = 0 # 目前加到數字多少 0 > 1 > 2
i_count = 0 # 計算達成 數字小於101 而且 是偶數的數字
while i_number < 101:
      if i_number%2 == 0:
         i_total += i_number
         i_count += 1
      i_number += 1
print(f"總和為{i_total},加總倒第{i_count}個數字")    
```

練習題 2：倒扣挑戰 — 幾次後變成負數？
請從 100 開始倒扣，每次減掉一個遞增的數字（第一次減 1，第二次減 2，第三次減 3...），請問總共扣了幾次才會讓總和變成負的？以及最後總和是多少？

範例邏輯：

第一次：100 - 1 = 99

第二次：99 - 2 = 97

...

當總和小於 0，就停止

```python
score = 100 #初始分數
step = 1 #一開始的數字 1 > 2 >3
total_deducted = 0 #計算step總共加總多少

while score > 0:
    score -= step
    total_deducted += step
    step += 1

print(f"扣了 {step - 1} 次")
print(f"最後扣掉 {step - 1} 分")
print(f"總共扣了 {total_deducted} 分")
print(f"扣完後分數為 {score}")
```

```python
score = 100
step = 1
total_deducted = 0

while score > 0:
    print(f"第 {step} 次扣 {step} 分，剩下 {score - step} 分")
    score -= step
    total_deducted += step
    step += 1
```
