字典是以「鍵（key）」與「值（value）」成對儲存資料的結構。常用於儲存具名稱或標籤的資訊，例如學生資料、商品清單等。

```python
person = {
    "name": "Alice",
    "age": 25,
    "is_student": True
}
```

✅ 建立字典
✅ 取值（使用 key）
✅ 新增或修改資料
✅ 檢查某 key 是否存在

```python
student = {
    "name": "Tom",
    "score": 88
}
```

```python
print(student["name"])   # Tom
print(student["score"])  # 88
```

```python
student["score"] = 90           # 修改
student["gender"] = "male"      # 新增
print(student["score"])
```

```python
if "score" in student:
    print("有分數欄位")
```

# 字典迴圈
你可以用 for 迴圈來走訪整個字典：
# 或使用 .items() 一次取得 key 和 value：

```python
for key in student:
    print(key, ":", student[key])
```

```python
for key, value in student.items():
    print(f"{key} 是 {value}")
```

巢狀 Dictionary（字典中包含字典）
在 Python 中，字典的值（value）也可以是另一個字典，這樣的結構稱為「巢狀 dictionary（Nested Dictionary）」，可用來表示更複雜或結構化的資料。

```python
students = {
    "001": {
        "name": "Alice",
        "email": "alice@example.com",
        "score": 90
    },
    "002": {
        "name": "Bob",
        "email": "bob@example.com",
        "score": 85
    },
    "003": {
        "name": "AZA",
        "email" : "aza@gmail.com",
        "score" : 87
    }
}
```

存取巢狀 Dictionary 的資料
要存取巢狀資料，使用連續的中括號 []：

```python
print(students["001"]["name"])   # 輸出 Alice
print(students["002"]["score"])  # 輸出 85
print(students["003"]["email"])  # 輸出AZA@gmail.com
```

🔹 修改巢狀 Dictionary 的資料
可以直接指定要修改的欄位：

```python
students["001"]["score"] = 95
```

🔹 巢狀迴圈列出所有資料

```python
for student_id, info in students.items():
    print("學號：", student_id)
    for key, value in info.items():
        print(f"  {key} : {value}")
```
