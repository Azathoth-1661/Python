建立一個學生清單（students），每個學生是一個字典，包含：

id：學號（字串）

name：名字（字串）

score：成績（整數）

範例資料格式如下（請自己輸入 3 個學生）：

使用 for 迴圈印出所有學生的資訊，例如：

學號: 001, 姓名: Amy, 成績: 90
學號: 002, 姓名: Bob, 成績: 85

```python
students = [
    {"id": "001", "name": "Amy", "score": 90},
    {"id": "002", "name": "Bob", "score": 85},
    {"id": "003", "name": "Charlie", "score": 78}
]

for students_total in students:
    print("學號:", students_total["id"], "姓名:", students_total["name"], "成績:", students_total["score"])
```

在這裡，你正確使用了 for students_total in students_s:，這代表你每次取得一位學生的資料。

但你卻用 students_s["id"]，這會導致錯誤，因為 students_s 是整個串列，而不是單一學生。

正確應該取用 students_total["id"]（因為 students_total 是那一位學生的字典）。

```python
def students_information(students_s):
    for students_total in students_s:
        print("學號:", students_total["id"], "姓名:", students_total["name"], "成績:", students_total["score"])
students_information(students)
```

```python
# 找出所有成績大於 80 分的學生，並印出他們的「學號」、「姓名」、「成績」
students = [
    {"id": "001", "name": "Amy", "score": 85},
    {"id": "002", "name": "Bob", "score": 77},
    {"id": "003", "name": "Charlie", "score": 92}
]

for students_total in students:
    if students_total["score"] > 80:
       print("學號:", students_total["id"], "姓名:", students_total["name"], "成績:", students_total["score"])
```

```python
# 把所有分數大於 80 的學生，存進一個新串列 high_scores

high_scores = []

for students_total in students:
    if students_total["score"] > 80:
        high_scores.append(students_total)

for student in high_scores:
    print("學號:", student["id"], "姓名:", student["name"], "成績:", student["score"])
```

```python
# 計算總分分
# 用 for 迴圈走訪 high_scores 
# 每個元素是學生字典
# 可以用 student["score"] 取出分數，加到 total_score 上。
total_score = 0
# 先將 total_score 初始化為 0。
for student in high_scores: # 使用 for 迴圈走訪 high_scores 串列中的每位學生。
    total_score += student["score"] # 將每位學生的分數加總到 total_score。
print("總分是:", total_score) # 最後在迴圈外印出總分。
```

```python
# 計算平均分
if len(high_scores) != 0: # 擔心 len(high_scores) 是 0 可以加個 if 判斷來保險。
   average = total_score / len(high_scores)
print("平均分數為:", round(average, 2))
```

```python
print(f"高分學生共 {len(high_scores)} 人，平均分數為：{round(average, 2)} 分")
```
