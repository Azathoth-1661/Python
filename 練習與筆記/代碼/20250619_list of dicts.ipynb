函式化

```python
def analyze_students(students_list):
    pass_students = []
    fail_students = []
    total_score = 0
    max_score = -1
    top_student = None

    for student in students_list:  # 這裡的 students_list 就是你要處理的學生資料（不限定是什麼變數名）
        if student["score"] >= 60:
            pass_students.append(student)
            total_score += student["score"]
        else:
            fail_students.append(student)

        # 找最高分的學生
        if student["score"] > max_score:
            max_score = student["score"]
            top_student = student

    # 計算平均分數（避免除以 0）
    average = total_score / len(pass_students) if len(pass_students) > 0 else 0

    # 結果輸出
    print("✅ 合格學生清單：")
    for student in pass_students:
        print("學號:", student["id"], "姓名:", student["name"], "成績:", student["score"])

    print("❌ 不合格學生清單：")
    for student in fail_students:
        print("學號:", student["id"], "姓名:", student["name"], "成績:", student["score"])

    print("📊 合格人數：", len(pass_students), "｜ 不合格人數：", len(fail_students))
    print("📈 合格學生平均分數：", round(average, 2))
    print("🏆 最高分學生：", top_student["id"], top_student["name"], top_student["score"])
```

```python
analyze_students(students_v2) 
# 把變數 students_v2 裡的資料「傳給了」函式中的 students_list
# 函式內的變數名稱（students_list）就像是暫時的代號
# 不管你傳進來的是 students_v2、students_data、data_list，只要在函式呼叫時傳進去，它就會自動對應到這個名字
```
