英制與公制轉換

```python
value = float(input('長度: '))
unit = input('單位: ')
if unit == 'in' or unit == '英寸':
    print('%f英寸 = %f公分' % (value, value * 2.54))
elif unit == 'cm' or unit == '公分':
    print('%f公分 = %f英寸' % (value, value / 2.54))
else:
    print('單位錯誤')
```

百分制成績轉換為等級制

```python
score = float(input('成绩: '))
if score >= 90:
    grade = 'A'
elif score >= 80:
    grade = 'B'
elif score >= 70:
    grade = 'C'
elif score >= 60:
    grade = 'D'
else:
    grade = 'E'
print('等級是:', grade)
```
