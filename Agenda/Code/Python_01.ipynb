數據類型轉換函數

```python
x = int(1)   
y = int(2.8) 
z = int("3") 
print(x)
print(y)
print(z)
print(type(x))
print(type(y))
print(type(z))
print("==========")
x = float(1)     
y = float(2.8)   
z = float("3")   
w = float("4.2") 
print(x)
print(y)
print(z)
print(w)
print(type(x))
print(type(y))
print(type(z))
print(type(w))
print("==========")
x = str("s1") 
y = str(2)    
z = str(3.0)  
print(x)
print(y)
print(z)
print(type(x))
print(type(y))
print(type(z))
```

鍵盤輸入簡單運算

```python
a = int(input('a = '))
b = int(input('b = '))
print('%d + %d = %d' % (a, b, a + b))
print('%d - %d = %d' % (a, b, a - b))
print('%d * %d = %d' % (a, b, a * b))
print('%d / %d = %f' % (a, b, a / b))
print('%d // %d = %d' % (a, b, a // b))
print('%d %% %d = %d' % (a, b, a % b))
print('%d ** %d = %d' % (a, b, a ** b))
```

華氏溫度轉攝氏溫度

```python
f = float(input('華氏溫度: '))
c = (f - 32) / 1.8
print('%.1f華氏溫度 = %.1f攝氏溫度' % (f, c))
```

使用圓半徑計算圓周長與面積

```python
radius = float(input('圓的半徑: '))
perimeter = 2 * 3.1416 * radius
area = 3.1416 * radius * radius
print('周長: %.2f' % perimeter)
print('面積: %.2f' % area)
```

計算是否為閏年

```python
year = int(input('年份: '))
is_leap = year % 4 == 0 and year % 100 != 0 or year % 400 == 0
print(is_leap)
```
