使用for-in 循環求1~100的總和

```python
sum = 0
for x in range(101):
    sum += x
print(sum)
```

使用for-in 循環求1~100的偶數總和

```python
sum = 0
for x in range(2, 101, 2):
    sum += x
print(sum)
```

猜數字

```python
import random

answer = random.randint(1, 100)
counter = 0
while True:
    counter += 1
    number = int(input('輸入數字: '))
    if number < answer:
        print('太小了')
    elif number > answer:
        print('太大了')
    else:
        print('猜對了')
        break
print('總共猜了%d次' % counter)
```

九九乘法表

```python
for i in range(1, 10):
    for j in range(1, i + 1):
        print('%d*%d=%d' % (i, j, i * j), end='\t')
    print()
```

判斷質數

```python
from math import sqrt

num = int(input('輸入一個正整數: '))
end = int(sqrt(num)) #平方根
is_prime = True
for x in range(2, end + 1):
    if num % x == 0:
        is_prime = False
        break
if is_prime and num != 1:
    print('%d是質數' % num)
else:
    print('%d不是質數' % num)
```

最大公約數與最小公倍數

```python
x = int(input('x = '))
y = int(input('y = '))

if x > y: #x大於y 則交換兩者的值
    x, y = y, x
for factor in range(x, 0, -1):
    if x % factor == 0 and y % factor == 0:
        print('%d和%d的最大公約數是%d' % (x, y, factor))
        print('%d和%d的最小公倍數是%d' % (x, y, x * y // factor))
        break
```
