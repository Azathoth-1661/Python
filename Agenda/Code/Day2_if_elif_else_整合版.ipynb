## 📘 重點語法整理

### 🔹 比較運算子
| 運算子 | 說明        | 範例             |
|--------|-------------|------------------|
| `==`   | 等於        | `a == b`         |
| `!=`   | 不等於      | `a != b`         |
| `>`    | 大於        | `a > b`          |
| `<`    | 小於        | `a < b`          |
| `>=`   | 大於等於    | `a >= b`         |
| `<=`   | 小於等於    | `a <= b`         |

### 🔹 邏輯運算子
| 運算子 | 說明                    | 範例                     |
|--------|-------------------------|--------------------------|
| `and`  | 且，兩條件皆為 True     | `a > 0 and b < 10`       |
| `or`   | 或，有一為 True 即成立 | `a > 0 or b < 10`        |
| `not`  | 反轉布林值              | `not a > 0`              |

練習題1. 簡易成績判斷

請使用者輸入一個數字分數，若分數大於等於 60，印出「及格」，否則印出「不及格」。

```python
try:
    score_number = int(input("請輸入你的成績"))
    if score_number >= 60:
       print("及格")
    else:
        print("不及格")
except ValueError:
    print("請輸入整數")
```

改良版 使用簡潔的條件與判斷 

Python isdigit()方法

Python isdigit() 方法检测字符串是否只由数字组成，只对 0 和 正数有效。

```python
score = input("請輸入你的成績：")
if score.isdigit():
    print("及格" if int(score) >= 60 else "不及格")
else:
    print("請輸入整數")
```

練習 2：分級成績顯示

請使用者輸入成績，依照以下邏輯顯示等級：

90 分以上：A

80~89 分：B

70~79 分：C

60~69 分：D

60 以下：F

```python
try:
    score = int(input("請輸入你的成績："))
    if score < 0 or score > 100: #檢查合理範圍內的數字
        print("請輸入 0 到 100 之間的數字")
    elif score >= 90:
        print("A")
    elif score >= 80: #不需要再檢查>90 因為上一層已經檢查了
        print("B")
    elif score >= 70:
        print("C")
    elif score >= 60:
        print("D")
    else:
        print("F")
except ValueError:
    print("請輸入正確的整數")
```

改良版 將條件簡化與數字合理性一起考量

```python
score = input("請輸入你的成績：")
if score.isdigit():
    score = int(score)
    if 0 <= score <= 100: #判斷沒有小於0跟大於100分的正常情況
        if score >= 90:
            print("A")
        elif score >= 80:
            print("B")
        elif score >= 70:
            print("C")
        elif score >= 60:
            print("D")
        else:
            print("F")
    else:
        print("請輸入 0~100 之間的數字")
else:
    print("請輸入整數")
```

額外挑戰題：判斷是否為閏年
請撰寫一段程式，讓使用者輸入一個年份，並根據以下規則判斷是否為閏年：

✅ 閏年判斷邏輯：
可以被 4 整除但不能被 100 整除，或

可以被 400 整除

🧾 範例輸入/輸出：
輸入：2000 → 輸出：是閏年

輸入：1900 → 輸出：不是閏年

輸入：2024 → 輸出：是閏年



```python
try:
    year_1 = int(input("請輸入年分"))
    if year_1 % 4 == 0 and year_1 % 100 != 0 or year_1 % 400 ==0:
       print("是閏年")
    else:
        print("不是閏年")
except ValueError:
    print("請重新輸出年分")    
```

改良版 封裝為函式並提升可讀性

```python
def is_leap(year):
    return(year %4 ==0 & year %100 !=0 ) or year %400 ==0

try:
    year = int(input("請輸入年分"))
    print("是閏年" if is_leap(year) else "不是閏年")

except ValueError:
    print("請輸入正確的數字")
```
