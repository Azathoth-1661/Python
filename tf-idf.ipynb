```python
txt = pd.read_excel("data.xlsx")
txt = str(txt)
words = jieba.lcut(txt)
new={}
for word in words:
    if len(word)==1:
        continue
    elif word == "狗狗漏" or word == "狗狗" or word =="gogoro":
        rword="Gogoro"
    else:
        rword=word
    # 對原來的字典new中判斷是否有rword這個人
    # 賦予rword鍵有就在原來rword所對應的值加1
    new[rword]=new.get(rword,0)+1
exc={"您好","請問","沒有","我們","目前","最近","感謝","..."}
for word in exc:
    del new[word]
item=list(new.items())
item.sort(key=lambda x:x[1],reverse=True)
for i in range(30):
    word,count=item[i]
    print("{0:<5}{1:>5}".format(word,count))
```
